```python
import numpy as np
import matplotlib.pyplot as plt
```

Advanced Activation Function Engien

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, x * alpha)
```

Activation Space Canvas Plotting

```python
x_range = np.linspace(-10, 10, 500)

y_sigmoid = sigmoid(x_range)
y_tanh = tanh(x_range)
y_relu = relu(x_range)
y_leaky = leaky_relu(x_range)

plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(x_range, y_sigmoid, color='blue', linewidth=2)
plt.title("Sigmoid Activation Space")
plt.grid(True, linestyle='--', alpha=0.5)

plt.subplot(2, 2, 2)
plt.plot(x_range, y_tanh, color='orange', linewidth=2)
plt.title("Hyperbolic Tangent (Tanh) Space")
plt.grid(True, linestyle='--', alpha=0.5)

plt.subplot(2, 2, 3)
plt.plot(x_range, y_relu, color='green', linewidth=2)
plt.title("ReLU Activation Space")
plt.grid(True, linestyle='--', alpha=0.5)

plt.subplot(2, 2, 4)
plt.plot(x_range, y_leaky, color='red', linewidth=2)
plt.title("Leaky ReLU Activation Space")
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

The Core Loss Functions Mechanics Engine

```python
def binary_cross_entropy(y_true, y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

Loss Space Canvas Rendering

```python
predictions_space = np.linspace(0.01, 0.99, 100)
targets_zero = np.zeros(100)
targets_one = np.ones(100)

bce_loss_y0 = [binary_cross_entropy(0, p) for p in predictions_space]
bce_loss_y1 = [binary_cross_entropy(1, p) for p in predictions_space]

mse_loss_y0 = [mean_squared_error(0, p) for p in predictions_space]
mse_loss_y1 = [mean_squared_error(1, p) for p in predictions_space]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(predictions_space, bce_loss_y1, color='blue', linewidth=2)
plt.plot(predictions_space, bce_loss_y0, color='red', linewidth=2)
plt.title("Binary Cross-Entropy Loss Space")
plt.xlabel("Predicted Probability")
plt.ylabel("Loss Cost")
plt.grid(True, linestyle='--', alpha=0.5)

plt.subplot(1, 2, 2)
plt.plot(predictions_space, mse_loss_y1, color='blue', linewidth=2)
plt.plot(predictions_space, mse_loss_y0, color='red', linewidth=2)
plt.title("Mean Squared Error (MSE) Space")
plt.xlabel("Predicted Value")
plt.ylabel("Loss Cost")
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
